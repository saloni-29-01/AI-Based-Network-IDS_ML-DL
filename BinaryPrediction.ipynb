# Binary Intrusion Detection on NSL-KDD (normal vs attack)

> **Research notebook, preserved from the original project (Mohammed Ali Cheddad).** The production pipeline used by the IDS application lives in `app/` (`app/models/trainer.py`); it does not depend on running these cells. Run `python scripts/train_models.py` to train the application's models.

## Objective
Classify each NSL-KDD connection record as **normal** or **attack** and compare classical ML models (Logistic Regression, Naive Bayes, Linear SVM, Decision Tree, Random Forest) with a neural network.

## Preprocessing
Robust scaling of numeric features, one-hot encoding of `protocol_type`, `service` and `flag`, and attack labels collapsed to a binary outcome.

## Models
Classical scikit-learn models, then a Keras network with a RandomizedSearchCV over its hyperparameters. Note: the "CNN" in this notebook is a dense multilayer perceptron; the convolutional model is in `MulticlassPrediction.ipynb` and `app/models/cnn.py`.

## Evaluation and results: read this caveat
This notebook evaluates on a **random 80/20 split of KDDTrain+**, and fits the scaler and PCA before the split. Scores of about 99% here therefore **overstate** generalisation: the official KDDTest+ split contains attack types that never appear in training. The application evaluates every model on KDDTest+ and KDDTest-21 with preprocessing fitted on training data only; see `docs/RESULTS.md` (binary accuracy on KDDTest+ is roughly 75-82%).

## Preprocessing the data

```python
def Scaling(df_num, cols):
    std_scaler = RobustScaler()
    std_scaler_temp = std_scaler.fit_transform(df_num)
    std_df = pd.DataFrame(std_scaler_temp, columns =cols)
    return std_df
```

```python
cat_cols = ["is_host_login", "protocol_type", "service", "flag", "land",
            "logged_in", "is_guest_login", "level", "outcome"]


def preprocess(dataframe):
    """Scale numeric columns, drop bad rows/duplicates, binarize the
    outcome label, and one-hot encode the categorical columns."""
    df_num = dataframe.drop(cat_cols, axis=1)
    num_cols = df_num.columns
    scaled_df = Scaling(df_num, num_cols)

    dataframe.drop(labels=num_cols, axis="columns", inplace=True)
    dataframe[num_cols] = scaled_df[num_cols]
    dataframe.dropna(inplace=True)
    dataframe.drop_duplicates(inplace=True)

    # Build the binary target as a fresh numeric column in one step rather
    # than assigning ints into the existing string column via .loc: pandas
    # now defaults to a strict StringDtype for object/text columns, which
    # rejects mixed str/int assignment (the original .loc pattern raised
    # `TypeError: Invalid value '0' for dtype 'str'` under pandas 2.x/3.x).
    dataframe["outcome"] = np.where(dataframe["outcome"] == "normal", 0, 1)

    dataframe = pd.get_dummies(
        dataframe, columns=["protocol_type", "service", "flag"], dtype=int
    )

    return dataframe
```

## ***Evaluate ML Models Function***

```python
from sklearn import metrics


def evaluate_classification_RF(model, name, X_train, X_test, y_train, y_test):
    """Print accuracy/precision/recall/F1 and plot confusion matrices for a
    scikit-learn classifier (used for every non-neural-network model below)."""
    # Predictions
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    # Compute confusion matrix for both train and test sets
    train_confusion_matrix = metrics.confusion_matrix(y_train, train_pred)
    test_confusion_matrix = metrics.confusion_matrix(y_test, test_pred)

    # Calculate metrics
    train_accuracy = metrics.accuracy_score(y_train, train_pred)
    train_precision = metrics.precision_score(y_train, train_pred)
    train_recall = metrics.recall_score(y_train, train_pred)
    train_f1 = metrics.f1_score(y_train, train_pred)

    test_accuracy = metrics.accuracy_score(y_test, test_pred)
    test_precision = metrics.precision_score(y_test, test_pred)
    test_recall = metrics.recall_score(y_test, test_pred)
    test_f1 = metrics.f1_score(y_test, test_pred)

    # Print metrics
    print(f"Training Metrics for {name}:")
    print(f"Accuracy: {train_accuracy*100:.2f}%, Precision: {train_precision*100:.2f}%, "
          f"Recall: {train_recall*100:.2f}%, F1 Score: {train_f1*100:.2f}%")
    print(f"Testing Metrics for {name}:")
    print(f"Accuracy: {test_accuracy*100:.2f}%, Precision: {test_precision*100:.2f}%, "
          f"Recall: {test_recall*100:.2f}%, F1 Score: {test_f1*100:.2f}%")

    # Plot confusion matrix
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    sns.heatmap(train_confusion_matrix, annot=True, fmt="g")
    plt.title(f"Training Confusion Matrix for {name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.subplot(1, 2, 2)
    sns.heatmap(test_confusion_matrix, annot=True, fmt="g")
    plt.title(f"Testing Confusion Matrix for {name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.show()
```

## ***Model Evaluation***

```python
import matplotlib.pyplot as plt
from sklearn import metrics
import seaborn as sns

def evaluate_classification(model, name, X_train, X_test, y_train, y_test):
    # Convert probabilities to binary labels
    train_pred = (model.predict(X_train) > 0.5).astype("int32")
    test_pred = (model.predict(X_test) > 0.5).astype("int32")

    # Calculate metrics for training data
    train_accuracy = metrics.accuracy_score(y_train, train_pred)
    train_precision = metrics.precision_score(y_train, train_pred)
    train_recall = metrics.recall_score(y_train, train_pred)
    train_f1 = metrics.f1_score(y_train, train_pred)

    # Calculate metrics for testing data
    test_accuracy = metrics.accuracy_score(y_test, test_pred)
    test_precision = metrics.precision_score(y_test, test_pred)
    test_recall = metrics.recall_score(y_test, test_pred)
    test_f1 = metrics.f1_score(y_test, test_pred)

    # Print metrics
    print(f"Training Metrics for {name}:")
    print(f"Accuracy: {train_accuracy*100:.2f}%, Precision: {train_precision*100:.2f}%, Recall: {train_recall*100:.2f}%, F1 Score: {train_f1*100:.2f}%")
    print(f"Testing Metrics for {name}:")
    print(f"Accuracy: {test_accuracy*100:.2f}%, Precision: {test_precision*100:.2f}%, Recall: {test_recall*100:.2f}%, F1 Score: {test_f1*100:.2f}%")

    # Confusion matrices
    train_confusion_matrix = metrics.confusion_matrix(y_train, train_pred)
    test_confusion_matrix = metrics.confusion_matrix(y_test, test_pred)

    # Plot confusion matrix for training data
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    sns.heatmap(train_confusion_matrix, annot=True, fmt='g')
    plt.title(f'Training Confusion Matrix for {name}')
    plt.xlabel('Predicted Labels')
    plt.ylabel('True Labels')

    # Plot confusion matrix for testing data
    plt.subplot(1, 2, 2)
    sns.heatmap(test_confusion_matrix, annot=True, fmt='g')
    plt.title(f'Testing Confusion Matrix for {name}')
    plt.xlabel('Predicted Labels')
    plt.ylabel('True Labels')
    plt.show()


evaluate_classification(model, "CNN", x_train, x_test, y_train, y_test)
```

## ***Optimisation Using RandomizedSearchCV***

```python
import tensorflow as tf
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import RandomizedSearchCV
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.models import Sequential


def build_model(units1=64, units2=32, learning_rate=0.001):
    """Build a small feed-forward binary classifier for hyperparameter search."""
    model = Sequential([
        Input(shape=(x_train.shape[1],)),
        Dense(units=units1, activation="relu"),
        Dense(units=units2, activation="relu"),
        Dense(1, activation="sigmoid"),
    ])
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
        loss="binary_crossentropy",
        metrics=["accuracy"],
    )
    return model


# Define the hyperparameter search space.
# NOTE (TF 2.11+ compatibility): `tensorflow.keras.wrappers.scikit_learn`
# was removed from TensorFlow; SciKeras is its maintained replacement.
# SciKeras routes keyword arguments through to `build_model` via a
# `model__` prefix, so the parameter names below must match its signature.
param_grid = {
    "model__units1": [64, 128, 256, 512],
    "model__units2": [32, 64, 128, 256],
    "model__learning_rate": [0.01, 0.001, 0.0001],
    "batch_size": [16, 32, 64],
    "epochs": [10, 20, 30],
}

# TensorBoard callback for tracking search runs
tensorboard_callback = tf.keras.callbacks.TensorBoard(log_dir="./logs")

keras_clf = KerasClassifier(model=build_model, verbose=0, callbacks=[tensorboard_callback])

random_search = RandomizedSearchCV(
    estimator=keras_clf, param_distributions=param_grid, n_iter=10, cv=3, n_jobs=1
)

# Use a GPU automatically when one is available, otherwise fall back to
# CPU so the notebook still runs on machines without a GPU.
device = "/GPU:0" if tf.config.list_physical_devices("GPU") else "/CPU:0"
with tf.device(device):
    random_search_results = random_search.fit(x_train, y_train)

print("Best: %f using %s" % (random_search_results.best_score_, random_search_results.best_params_))
```
